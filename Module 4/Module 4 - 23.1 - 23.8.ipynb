```python
# Module 4 - Exercises 23.1 - 23.8

import csv

# Exercise 23.1
content = '''author,book
J R R Tolkien,The Hobbit
Lynne Truss,"Eats, Shoots & Leaves"
'''

with open('books.csv','w',encoding='utf-8') as file:
    file.write(content)
```

```python
# Exercise 23.2
with open('books.csv','rt',encoding='utf-8') as file:
    books_reader = csv.DictReader(file)
    books = [row for row in books_reader]

for book in books:
    print(book)
```

```python
# Exercise 23.3
content2 = '''title,author,year
The Weirdstone of Brisingamen,Alan Garner,1960
Perdido Street Station,China Miéville,2000
Thud!,Terry Pratchett,2005
The Spellman Files,Lisa Lutz,2007
Small Gods,Terry Pratchett,1992'''

with open('books2.csv','wt',encoding='utf-8') as file:
    file.write(content2)
```

```python
# Exercise 23.4

import sqlite3

connection = sqlite3.connect('books.db')
cursor = connection.cursor()

cursor.execute('''CREATE TABLE IF NOT EXISTS book
               (title TEXT, author TEXT, year INTEGER)''')
```

```python
# Exercise 23.5

with open('books2.csv','r',encoding='utf-8') as file:
    fin = csv.DictReader(file)
    rows = [(row['title'], row['author'],row['year']) for row in fin]

cursor.executemany('INSERT INTO book VALUES (?,?,?)', rows)
connection.commit()
```

```python
# Exercise 23.6

cursor.execute('SELECT title FROM book ORDER BY title')
for row in cursor.fetchall():
    print(row[0])
```

```python
# Exercise 23.7

cursor.execute("SELECT * FROM book ORDER BY year")
```

```python
# Exercise 23.8

import sqlalchemy

engine = sqlalchemy.create_engine('sqlite:///books.db')

with engine.connect() as connection:
    query = sqlalchemy.text("SELECT title FROM book ORDER BY title")
    result = connection.execute(query)

    for row in result:
        print(row[0])
```
